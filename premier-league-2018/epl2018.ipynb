<h2>12. Retrieve, Calculate, Analyse and Sort the Data (Premier League Standings)</h2>
<p>The data that we retrieved are already in a structured form but we can further organise them so that more meaningful information can be produced. Previously, we cannot determine the number of goal scored/conceded, total points collected and the number of win/draw/loss because the data are not organised properly. For example, the dataset does not contain any information about the total points collected by every team and it only provides us with the data such as 'Full Time Result'. Based on the data given, we need to perform some calculations on our own and determine the total points received by teams. Once we have determined the total points, then we will be able to rank teams based on the total points they obtained.<br><br>
Points System:<br>
    <ul>
        <li>Three (3) points will be awarded for a win</li>
        <li>One (1) point will be awarded for a draw</li>
        <li>No point (0) will be awarded for a loss</li>
    </ul>
Equation for Total Points:
<pre>
    Total Points = (Total Win &#215; 3) + (Total Draw &#215; 1)
</pre></p>

```python
pl_teams = epl2018_df["HomeTeam"][epl2018_df["AwayTeam"].isin(epl2018_df["HomeTeam"])].drop_duplicates()
pl_standings = pd.DataFrame([])

for team in pl_teams:
    won = epl2018_df.loc[(epl2018_df['HomeTeam'] == team) & (epl2018_df['FTR'] == 'H')].FTR.count()
    won += epl2018_df.loc[(epl2018_df['AwayTeam'] == team) & (epl2018_df['FTR'] == 'A')].FTR.count()
    lost = epl2018_df.loc[(epl2018_df['HomeTeam'] == team) & (epl2018_df['FTR'] == 'A')].FTR.count()
    lost += epl2018_df.loc[(epl2018_df['AwayTeam'] == team) & (epl2018_df['FTR'] == 'H')].FTR.count()
    drawn = 38 - (won + lost)
    goals_for = epl2018_df.loc[(epl2018_df['HomeTeam'] == team)].FTHG.sum()
    goals_for += epl2018_df.loc[(epl2018_df['AwayTeam'] == team)].FTAG.sum()
    goals_against = epl2018_df.loc[(epl2018_df['HomeTeam'] == team)].FTAG.sum()
    goals_against += epl2018_df.loc[(epl2018_df['AwayTeam'] == team)].FTHG.sum()
    goal_difference = goals_for - goals_against
    points = (won * 3) + (drawn * 1)
    pl_standings = pl_standings.append(pd.DataFrame({'Club': team, 'Played': 38, 'Won': won, 'Drawn': drawn, 
                                                     'Lost': lost, 'GF': goals_for, 'GA': goals_against, 'GD': goal_difference, 
                                                     'Points': points}, index = [1]))

pl_standings = pl_standings.sort_values(by=['Points', 'GD', 'GF'], ascending=False) 
pl_standings.index = np.arange(1,len(pl_standings)+1)
pl_standings.to_csv('assets\csv_files\pl_standings.csv', index=False)
pl_standings
```

<h2>13. Retrieve, Calculate, Analyse and Sort the Data (Probability)</h2>
<p>Since we have organised the data, we can easily yield a lot of information from it. We want to find out which teams had the highest/lowest probabilities of win, draw and loss in the Premier League 2018-19. Table below represents the probabilities of every outcome for each Premier League team.<br><br>
Equation for probabilities:<br>
<pre>
    Probability of Win = Total of Win &#247; Total Games Played
    Probability of Draw = Total of Draw &#247; Total Games Played
    Probability of Loss = Total of Loss &#247; Total Games Played
</pre>
<i>Note: In total, there are 38 games to be played by a single team in the Premier League.</i>
</p>

```python
probabilities = pl_standings.loc[:, ['Club', 'Won', 'Drawn', 'Lost']]

probabilities['Won'] = (probabilities['Won'] / 38).round(2)
probabilities['Drawn'] = (probabilities['Drawn'] / 38).round(2)
probabilities['Lost'] = (probabilities['Lost'] / 38).round(2)
probabilities = probabilities.rename(columns={"Won" : "Win Probability", "Drawn" : "Draw Probability", 
                                              "Lost" : "Loss Probability"})
probabilities.to_csv('assets\csv_files\probabilities.csv', index=False)
probabilities
```

<h2>14. Retrieve, Calculate, Analyse and Sort the Data (Goal-to-Attempt Ratio)</h2>
<p>We want to know not only the number of goal scored but also which teams were more clinical in terms of finishing. Table below contains the total goal scored, total attempts made as well as conversion rate (in percentage) by every team. In the Premier League 2018-19, Arsenal were the most clinical, with 15.67 percent of the total attempts were successfully converted into goals.<br><br>
Equation for Conversion Rate (%):<br>
<pre>
    Conversion Rate = Goal Scored &#247; Total Attempts
</pre></p>

```python
goalscoring = pd.DataFrame([])

for team in pl_teams:
    total_attempts = epl2018_df.loc[(epl2018_df['HomeTeam'] == team)].HS.sum()
    total_attempts += epl2018_df.loc[(epl2018_df['AwayTeam'] == team)].AS.sum()
    goalscoring = goalscoring.append(pd.DataFrame({'Club': team, 'Goal Scored': pl_standings.loc[(pl_standings['Club'] == team)].GF, 
                                                   'Total Attempts': total_attempts, 'Conversion Rate (%)': pl_standings.loc[(pl_standings['Club'] == team)].GF / total_attempts * 100}, 
                                                    index = pl_standings.loc[(pl_standings['Club'] == team)].index.copy()))

goalscoring.sort_values(by='Conversion Rate (%)', ascending=False, inplace=True)
goalscoring['Conversion Rate (%)'] = goalscoring['Conversion Rate (%)'].round(2)
goalscoring.to_csv('assets\csv_files\goalscoring.csv', index=False)
goalscoring.head(5)
```

<h2>15. Retrieve, Calculate, Analyse and Sort the Data (Disciplinary Record)</h2>
<p><i>"Which teams were more disciplined and which teams were not?"</i><br><br>This question has made us feel curious to learn more and in the end has led us to analyse the data provided. Based on the total cards received, Liverpool were the most disciplined team, receiving only 39 cards in one whole season meanwhile Watford, with 81 cards received in total, were the most indisciplined side in the Premier League 2018-19.<br><br>
Equation for Avg Per Game (Y <i>or</i> R):
<pre>
    Avg Per Game (Y <i>or</i> R) = Total Yellow Cards <i>or</i> Total Red Cards &#247; Total Games Played
</pre>
Equation for Total Cards Received:
<pre>
    Total Cards Received = Total Yellow Cards + Total Red Cards
</pre>
<i>Note: Y represents Yellow Card meanwhile R means Red Card.</i></p>

```python
discipline = pd.DataFrame([])

for team in pl_teams:
    total_yellow = epl2018_df.loc[(epl2018_df['HomeTeam'] == team)].HY.sum()
    total_yellow += epl2018_df.loc[(epl2018_df['AwayTeam'] == team)].AY.sum()
    total_red = epl2018_df.loc[(epl2018_df['HomeTeam'] == team)].HR.sum()
    total_red += epl2018_df.loc[(epl2018_df['AwayTeam'] == team)].AR.sum()
    discipline = discipline.append(pd.DataFrame({'Club': team, 'Yellow Card (Y)': total_yellow, 'Avg Per Game (Y)': total_yellow / 38, 
                                                 'Red Card (R)': total_red, 'Avg Per Game (R)': total_red / 38, 'Total Cards Received': total_yellow + total_red}, 
                                                index = pl_standings.loc[(pl_standings['Club'] == team)].index.copy()))

discipline.sort_values(by=['Total Cards Received', 'Red Card (R)'], ascending=False, inplace=True)
discipline.to_csv('assets\csv_files\discipline.csv', index=False)
discipline.head(5)
```

<h2>16. Retrieve, Calculate, Analyse and Sort the Data (Points Collected Per Game)</h2>
<p>We further organised them by replacing the column "Full Time Result (FTR)" with a column called "Point Collected". The reason behind this is because the column "Full Time Result (FTR)" only indicates which team won, drawn or loss but it does not show exactly how many points were obtained and we want to visualise these data into graphs which can show the points collected on a game basis by every team. Thus, by replacing the data into more meaningful, we can easily represent the data into graphs.</p>

```python
collected_points = pd.DataFrame([])

for index, row in epl2018_df.iterrows():
    club = epl2018_df.loc[(epl2018_df.index == index)].HomeTeam
    date = epl2018_df.loc[(epl2018_df.index == index)].Date
    result = epl2018_df.loc[(epl2018_df.index == index)].FTR
    points = 0
    if result[index] == 'H':
        points = 3
    elif result[index] == 'D':
        points = 1
    collected_points = collected_points.append(pd.DataFrame({'Date': date, 'Club': club, "Point Collected": points}, 
                                                            index = [index]))
    club = epl2018_df.loc[(epl2018_df.index == index)].AwayTeam
    points = 0
    if result[index] == 'A':
        points = 3
    elif result[index] == 'D':
        points = 1
    collected_points = collected_points.append(pd.DataFrame({'Date': date, 'Club': club, "Point Collected": points}, 
                                                            index = [index]))

collected_points.index = np.arange(1,len(collected_points)+1)
collected_points.to_csv('assets\csv_files\collected_points.csv', index=False)
collected_points
```

<h2>17. Retrieve, Calculate, Analyse and Sort the Data (Goal Scored Per Game)</h2>
<p>We further organised them by combining the columns "Full Time Home Goal (FTHG)" and "Full Time Away Goal (FTAG)" into a new column called "Goals Scored" so it will be more meaningful and we can easily represent the data into graphs.</p>

```python
goal_history = pd.DataFrame([])

for index, row in epl2018_df.iterrows():
    club = epl2018_df.loc[(epl2018_df.index == index)].HomeTeam
    date = epl2018_df.loc[(epl2018_df.index == index)].Date
    goal = epl2018_df.loc[(epl2018_df.index == index)].FTHG
    goal_history = goal_history.append(pd.DataFrame({'Date': date, 'Club': club, "Goals Scored": goal}, index = [index]))
    club = epl2018_df.loc[(epl2018_df.index == index)].AwayTeam
    goal = epl2018_df.loc[(epl2018_df.index == index)].FTAG
    goal_history = goal_history.append(pd.DataFrame({'Date': date, 'Club': club, "Goals Scored": goal}, index = [index]))

goal_history.index = np.arange(1,len(goal_history)+1)
goal_history.to_csv('assets\csv_files\goal_history.csv', index=False)
goal_history
```
